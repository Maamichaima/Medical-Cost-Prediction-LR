```python
import pandas  as pd #Data manipulation
import numpy as np #Data manipulation
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['figure.figsize'] = [8,5]
plt.rcParams['font.size'] =14
plt.rcParams['font.weight']= 'bold'
# plt.style.use('seaborn-whitegrid')
```

```python
df = pd.read_csv("./insurance.csv");
print(f"num of rows and colums dataset ==> {df.shape}")
```

```python
df.head(10)
```

```python
sns.lmplot(df, x="bmi", y="charges", aspect=2, height=6)
plt.title('charges vs bmi')
```

```python
df.describe()
```

```python
plt.figure(figsize=(12, 4))
print(df.isnull())
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Missing value in the dataset');
plt.show()
```

```python
numerical_cols = ['age', 'bmi', 'children', 'charges']
df_numerical = df[numerical_cols]

corr = df_numerical.corr()

plt.figure(figsize=(9, 7))
sns.heatmap(corr, cmap = 'Wistia', annot= True);
plt.title('Matrice de Corrélation des Variables Numériques')
plt.show()

```

```python
f = plt.figure(figsize=(12,4))

ax=f.add_subplot(121) 
sns.distplot(df['charges'], bins=50, color='g', ax=ax)
ax.set_title("insurance charges")

ax=f.add_subplot(122) 
sns.distplot(np.log10(df['charges']), bins=50, color='b', ax=ax)
ax.set_title("insurance charges in $log$ sacle")
```

```python
f = plt.figure(figsize=(12,4))

ax = f.add_subplot(121)
sns.violinplot(x='sex', y='charges', data=df, palette='Wistia')
ax.set_title("violin plot female male")


ax = f.add_subplot(122)
sns.violinplot(x='smoker', y='charges', data=df, palette='Wistia')
ax.set_title("violin plot female male")
```

```python
f = plt.figure(figsize=(12,4))
sns.boxplot(x='children', hue='sex', y='charges', data=df)
plt.title('Box plot of charges vs children');
```

```python
f = plt.figure(figsize=(12,4))
sns.violinplot(x='region', y='charges', hue='sex', data=df)
```

```python
plt.figure(figsize=(14,6))
sns.violinplot(x='region', y='charges',hue='sex',data=df,palette='rainbow',split=True)
plt.title('Violin plot of charges vs children');
```

```python
f = plt.figure(figsize=(14,6))

ax = f.add_subplot(121)
sns.scatterplot(x='age', y='charges',data=df, palette='magma',hue='smoker', ax=ax)
ax.set_title('Violin plot of charges vs children');

ax = f.add_subplot(122)
sns.scatterplot(x='bmi', y='charges',data=df, palette='magma',hue='smoker', ax=ax)
ax.set_title('Violin plot of charges vs children');
```

```python
categorical_columns = ['sex', 'sex', 'smoker', 'region']
df_encode = pd.get_dummies(data=df, columns=categorical_columns, drop_first=True)
print(df_encode)
```

```python
print('Columns in original data frame:\n', df.columns.values)
print('\nNumber of rows and columns in the dataset:', df.shape)
print('\nColumns in data frame after encoding dummy variable:\n', df_encode.columns.values)
print('\nNumber of rows and columns in the dataset:', df_encode.shape)
```
